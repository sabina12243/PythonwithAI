```python
import numpy as np
import matplotlib.pyplot as plt

n_values = [500, 1000, 2000, 5000, 10000, 15000, 20000, 50000, 100000]

for n in n_values:
    dice1 = np.random.randint(1, 7, n)
    dice2 = np.random.randint(1, 7, n)
    s = dice1 + dice2

    h, h2 = np.histogram(s, range(2, 14))

    plt.bar(h2[:-1], h / n)
    plt.title(f"Histogram of dice sums (n={n})")
    plt.xlabel("Sum of two dice")
    plt.ylabel("Relative frequency")
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

df = pd.read_csv("weight-height.csv")

print(df.head())

X = df[["Height"]]          
y = df["Weight"]            

plt.figure(figsize=(8, 6))
plt.scatter(X, y, alpha=0.4)
plt.title("Height vs Weight")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.show()

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

plt.figure(figsize=(8, 6))
plt.scatter(X, y, alpha=0.4, label="Data")
plt.plot(X, y_pred, linewidth=3, label="Regression line")
plt.title("Linear Regression Fit")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.legend()
plt.show()

rmse = np.sqrt(mean_squared_error(y, y_pred))
r2 = r2_score(y, y_pred)

print("RMSE:", rmse)
print("R²:", r2)

print("\nModel assessment:")
print("- The regression line should visually align with the upward trend in the scatter plot.")
print("- A higher R² indicates strong linear dependence.")
print("- Lower RMSE indicates smaller prediction error.")
```
